```python
# 1. Import the necessary libraries
import pandas as pd

# 2. Load the dataset
df = pd.read_csv('../data/raw/MLS.csv')

# 3. Display the first few rows
print(df.head())

# 4. Basic info about the dataset
print(df.info())

# 5. Quick statistics
print(df.describe())
```

```python
# Convert 'Date' from string to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Check if it worked
print(df.dtypes)
```

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
# Drop rows where 'CompIndex' is missing (and other main Composite columns)
df = df.dropna(subset=['CompIndex', 'CompBenchmark', 'CompYoYChange', 'SFDetachIndex', 'SFDetachBenchmark'])

# Check how many rows left
print(df.shape)
```

```python
# Rename columns for better clarity
df = df.rename(columns={
    'CompIndex': 'Composite_Index',
    'CompBenchmark': 'Composite_Benchmark_Price',
    'CompYoYChange': 'Composite_YoY_Change_Percent',
    'SFDetachIndex': 'Detached_Index',
    'SFDetachBenchmark': 'Detached_Benchmark_Price',
    'SFDetachYoYChange': 'Detached_YoY_Change_Percent',
    'SFAttachIndex': 'Attached_Index',
    'SFAttachBenchmark': 'Attached_Benchmark_Price',
    'SFAttachYoYChange': 'Attached_YoY_Change_Percent',
    'THouseIndex': 'Townhouse_Index',
    'THouseBenchmark': 'Townhouse_Benchmark_Price',
    'THouseYoYChange': 'Townhouse_YoY_Change_Percent',
    'ApartIndex': 'Apartment_Index',
    'ApartBenchmark': 'Apartment_Benchmark_Price',
    'ApartYoYChange': 'Apartment_YoY_Change_Percent'
})

# Verify new column names
print(df.columns)
```

```python
# Save the cleaned dataset
df.to_csv('../data/cleaned/cleaned_MLS.csv', index=False)

print("Cleaned dataset saved successfully!")
```
